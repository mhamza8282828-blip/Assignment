```python
def predict_student_score(age, gender, study, social, netflix, part_time_job, attendance, sleep, exercise, mental_health, extracurricular_participation):
    input_data = {
        'age': [age],
        'study_hours_per_day': [study],
        'social_media_hours': [social],
        'netflix_hours': [netflix],
        'attendance_percentage': [attendance],
        'sleep_hours': [sleep],
        'exercise_frequency': [exercise],
        'mental_health_rating': [mental_health],
        'gender': [gender],                                         # 0: Female, 1: Male, 2: Other
        'part_time_job': [part_time_job],                           # 0: No, 1: Yes
        'extracurricular_participation': [extracurricular_participation] # 0: No, 1: Yes
    }
    
    student_df = pd.DataFrame(input_data)

    student_df = student_df[lr.feature_names_in_]
 
    predicted_score = lr.predict(student_df)
    return predicted_score

print("--- Enter Student Details to Predict Exam Score ---")
print("(For Gender: Female=0, Male=1, Other=2 | For Job & Participation: No=0, Yes=1)\n")

user_age = int(input("Enter Age: "))
user_gender = int(input("Enter Gender (0/1/2): "))
user_study = float(input("Enter Study Hours Per Day: "))
user_social = float(input("Enter Social Media Hours: "))
user_netflix = float(input("Enter Netflix Hours: "))
user_job = int(input("Enter Part time Job (0/1): "))
user_attendance = float(input("Enter Attendance Percentage: "))
user_sleep = float(input("Enter Sleep Hours: "))
user_exercise = int(input("Enter Exercise Frequency per week: "))
user_mental = int(input("Enter Mental Health Rating out of 10: "))
user_participation = int(input("Enter Extracurricular Participation (0/1): "))

final_result = predict_student_score(
    age=user_age, 
    gender=user_gender,
    study=user_study, 
    social=user_social, 
    netflix=user_netflix,
    part_time_job=user_job,
    attendance=user_attendance, 
    sleep=user_sleep, 
    exercise=user_exercise, 
    mental_health=user_mental,
    extracurricular_participation=user_participation
)
print(f" Predicted Exam Score: {round(final_result[0], 2)}")
```
